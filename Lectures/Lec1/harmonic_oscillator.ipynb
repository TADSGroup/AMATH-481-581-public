# Harmonic oscillator solution

Plots the analytic solution of the harmonic oscillator
$$
\frac{d^2 u}{dt^2} + \omega^2 u = 0, \qquad u(0) = u_0, \quad \dot u(0) = v_0,
$$
given by
$$
u(t) = u_0 \cos(\omega t) + \frac{v_0}{\omega}\sin(\omega t).
$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# parameters
omega = 2.0
u0 = 1.0
v0 = 0.5

t = np.linspace(0, 4 * np.pi / omega, 500)
u = u0 * np.cos(omega * t) + (v0 / omega) * np.sin(omega * t)
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(t, u)
ax.set_xlabel("$t$")
ax.set_ylabel("$u(t)$")
ax.set_title("Harmonic oscillator solution")
fig.tight_layout()
fig.savefig("harmonic_oscillator.png", dpi=300)
plt.show()
```
