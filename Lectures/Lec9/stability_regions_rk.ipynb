# Absolute stability regions of explicit Runge-Kutta methods

Companion notebook for `Lec9.tex`. Applying a one-step method to the test equation $u' = \lambda u$ gives
$u_{n+1} = R(z)\,u_n$ with $z = \lambda \Delta t$. The region of absolute stability is
$\{z \in \mathbb{C} : |R(z)| \le 1\}$. For the three methods considered here,
$$
\begin{aligned}
\text{forward Euler:} \quad & R(z) = 1 + z, \\
\text{explicit midpoint (RK2):} \quad & R(z) = 1 + z + \tfrac{z^2}{2}, \\
\text{classical RK4:} \quad & R(z) = 1 + z + \tfrac{z^2}{2} + \tfrac{z^3}{6} + \tfrac{z^4}{24}.
\end{aligned}
$$
Each is the Taylor polynomial of $e^z$ truncated at the order of the method.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
R = {
    "forward Euler": lambda z: 1 + z,
    "explicit midpoint (RK2)": lambda z: 1 + z + z**2 / 2,
    "classical RK4": lambda z: 1 + z + z**2 / 2 + z**3 / 6 + z**4 / 24,
}

x = np.linspace(-3.5, 1.0, 600)
y = np.linspace(-3.5, 3.5, 600)
X, Y = np.meshgrid(x, y)
Z = X + 1j * Y
```

```python
fig, ax = plt.subplots(figsize=(6, 6))
colors = ["C0", "C1", "C2"]
# draw the largest region first so the smaller ones remain visible on top
for (name, Rf), c in reversed(list(zip(R.items(), colors))):
    absR = np.abs(Rf(Z))
    ax.contourf(X, Y, absR, levels=[0, 1], colors=[c], alpha=0.3)
    ax.contour(X, Y, absR, levels=[1], colors=[c], linewidths=2)
    ax.plot([], [], color=c, linewidth=6, alpha=0.5, label=name)

ax.axhline(0, color="k", linewidth=0.5)
ax.axvline(0, color="k", linewidth=0.5)
ax.set_aspect("equal")
ax.set_xlabel(r"$\mathrm{Re}(z)$")
ax.set_ylabel(r"$\mathrm{Im}(z)$")
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles[::-1], labels[::-1], loc="upper center", bbox_to_anchor=(0.5, -0.1), ncol=2, frameon=False)
fig.tight_layout()
fig.savefig("stability_regions_rk.png", dpi=300, bbox_inches="tight")
plt.show()
```

The stability regions grow with the order of the method: on the negative real axis they extend to
$z = -2$ for forward Euler and RK2, and to $z \approx -2.785$ for RK4. RK4's region also contains a segment of the
imaginary axis, $|\mathrm{Im}(z)| \le 2\sqrt{2} \approx 2.83$, while forward Euler and RK2 touch the imaginary axis
only at the origin. All three regions are bounded, as is the case for any explicit Runge-Kutta method, since $R(z)$
is a polynomial.
